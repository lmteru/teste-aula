# Aula 4 - Monitoração

## 💡 O que é Monitoração?

**Monitoração** é o processo contínuo de **coletar, analisar e acompanhar métricas e logs** de um sistema, com o objetivo de garantir que ele esteja funcionando corretamente, com desempenho adequado e dentro dos padrões esperados.  

Monitorar é **observar o comportamento de um sistema em tempo real**, buscando **detectar anomalias, falhas ou degradações de performance**.

---

## 🚨 O que são Alertas?

**Alertas** são **notificações automáticas** geradas quando uma métrica monitorada **ultrapassa um limite definido** ou apresenta **um comportamento anômalo**.  

Esses alertas servem para **avisar equipes técnicas** (de dados, operações, segurança, etc.) de que algo precisa ser investigado ou corrigido.  

Exemplo de alertas:
- A latência da API ultrapassou 2 segundos  
- A acurácia do modelo caiu abaixo de 80%  
- O volume de requisições diárias reduziu drasticamente

Os alertas são fundamentais para **detecção precoce de problemas** e **resposta rápida**, evitando impactos no negócio.

---

## ⚙️ Monitoração de Aplicações

A **monitoração de aplicações tradicionais** foca em **aspectos técnicos e operacionais** da execução de software.  

Principais objetivos:
- Garantir **disponibilidade**, **performance** e **estabilidade** da aplicação.  
- Detectar erros, lentidão, travamentos ou sobrecarga de recursos.  

Métricas comuns:
- Latência e tempo de resposta  
- Taxa de erros (HTTP 500, 404, etc.)  
- Consumo de CPU, memória e disco  
- Tráfego de rede e número de requisições  
- Logs de eventos e exceções  

Ferramentas comuns:
- Grafana
- Datadog
- CloudWatch

---
## 🧾 Versionamento de Modelos

O **versionamento de modelos** é o processo de **controlar as versões de treinamento, dados, parâmetros e artefatos** usados em cada modelo.  
Ele garante **reprodutibilidade**, **rastreabilidade** e **governança**.

### 🔍 Por que versionar?

- Saber **qual modelo está em produção**  
- Rastrear **quais dados e hiperparâmetros** foram usados  
- Permitir **rollback** em caso de problemas  
- Ajudar na **auditoria e compliance**  

### 🧱 O que versionar?

1. **Código do modelo** (arquitetura, scripts de treino)  
2. **Dados usados no treinamento** (dataset e pré-processamento)  
3. **Pesos e parâmetros treinados**

# 🧠 Diferenças entre Monitoração de Aplicações e Monitoração de Modelos de Machine Learning


## 🤖 Monitoração de Modelos de Machine Learning

Já a **monitoração de modelos de Machine Learning** vai além do funcionamento técnico: ela busca garantir que o **modelo continue fazendo boas previsões ao longo do tempo**.  

Um modelo pode estar **rodando perfeitamente** (sem erros de sistema), mas **sendo cada vez menos preciso** devido a mudanças nos dados ou no comportamento do mundo real.

Principais objetivos:
- Detectar **degradação de desempenho** do modelo  
- Identificar **mudanças na distribuição dos dados** (*data drift*)  
- Acompanhar **mudanças na relação entre entrada e saída** (*concept drift*)

# 📉 Monitoramento e Manutenção de Modelos de Machine Learning

## 🧭 Introdução

Com o tempo, modelos de Machine Learning tendem a **perder desempenho** devido a mudanças nos dados, no ambiente ou no próprio comportamento dos usuários.  
Esse fenômeno é conhecido como **drift**, e monitorá-lo é essencial para manter a **confiabilidade e validade** das previsões.

---

## ⚙️ Tipos de Drift em Modelos de Machine Learning

### 1. 📦 Data Drift (Drift de Dados)

O **data drift** ocorre quando a **distribuição das variáveis de entrada (features)** muda ao longo do tempo, mesmo que a relação entre as entradas e as saídas permaneça a mesma.  

Essas mudanças podem afetar a capacidade do modelo de generalizar corretamente.

**Exemplo:**  
Um modelo de detecção de fraudes treinado com dados de cartões de crédito de 2022 pode começar a falhar em 2025, quando novos padrões de comportamento do consumidor aparecem.

**Causas comuns:**
- Mudanças sazonais  
- Alterações no comportamento do usuário  
- Novos produtos, regiões ou segmentos de dados  
- Erros de coleta ou transformação de dados  

**Sinais típicos:**
- Mudança de média, desvio padrão ou faixas de valores das features  
- Aumento no número de valores ausentes ou categóricos desconhecidos  

---

### 2. 🧠 Concept Drift (Drift de Conceito)

O **concept drift** acontece quando **a relação entre as variáveis de entrada (X) e a variável de saída (y)** muda com o tempo.  
Ou seja, **o significado do que o modelo tenta prever se altera**.

**Exemplo:**  
Um modelo que prevê cancelamento de clientes (*churn*) pode perder precisão quando a empresa muda suas políticas de preço — o que altera o que "significa" um cliente estar propenso a sair.

**Causas comuns:**
- Mudanças nas regras de negócio  
- Alterações em políticas públicas, economia, comportamento social  
- Mudanças nos padrões que definem a variável alvo  

**Sinais típicos:**
- Redução da acurácia, F1-score ou outras métricas de performance  
- Alterações na importância das variáveis  
- Diferença significativa entre previsões e valores reais (quando disponíveis)  

---

### 3. 🧩 Model Drift (Drift de Modelo)

O **model drift** é o resultado direto de *data drift* e *concept drift*.  
Refere-se à **degradação geral do desempenho preditivo** do modelo com o passar do tempo.

**Exemplo:**  
Um modelo de recomendação de produtos que gradualmente se torna menos relevante porque os gostos dos usuários mudaram ou novos produtos foram lançados.

**Como identificar:**
- Monitorar métricas de performance em produção  
- Comparar previsões recentes com dados rotulados de validação  
- Avaliar se o erro médio ou a incerteza aumentam com o tempo

## 📏 Métricas para Detecção de Drift

Monitorar *drifts* requer métricas estatísticas que quantifiquem **mudanças nas distribuições** dos dados ou nas relações entre variáveis.

### 🔹 Para Data Drift

Para detectar o Data drift é importante monitorar os dados de entrada, garantindo que a população de entrada ainda pertence a população dos dados que foram utilizados para treinar o modelo.

| Métrica | Descrição | Interpretação |
|----------|------------|----------------|
| **Kolmogorov–Smirnov (KS Test)** | Teste estatístico para comparar distribuições de duas amostras | Valores altos indicam mudança significativa |
| **Population Stability Index (PSI)** | Mede a estabilidade da distribuição de uma feature | PSI > 0.25 → possível drift |
| **Jensen–Shannon Divergence (JSD)** | Mede a divergência entre distribuições de probabilidade | Próximo de 0 → estável; maior → drift |
| **Chi-Square Test** | Avalia diferenças em distribuições categóricas | p-valor pequeno → mudança detectada |

### 🔹 Para Concept Drift

Para detectar o Concept Drift é importante monitorar a saída do modelo, garantindo que as métricas que foram definidas no momento do treinamento do modelo ainda estão sendo mantidas.

| Métrica | Descrição | Interpretação |
|----------|------------|----------------|
| **Acurácia / F1-score / ROC-AUC** | Comparação de métricas de desempenho ao longo do tempo | Queda consistente → sinal de concept drift |
| **Mean Prediction Drift** | Diferença média entre previsões antigas e novas | Aumentos súbitos → modelo instável |
| **Error Rate Stability** | Variação da taxa de erro entre janelas de tempo | Maior variação → possível concept drift |
